# 01 · Data: 1.5 m layers of each soil parameter, with the predicted P(Clay)

1. Reads the borehole file and keeps the **Downtown** boreholes (as V3).
2. For each target in `settings.toml → [cv] targets`, keeps the sample records with a value: record at most 3 m long, middle depth 0-15 m, value inside the target's `min`/`max`.
3. Gives every sample the classifier's **P(Clay)** at its 0.5 m cell: the **out-of-fold** classifier for development boreholes (fold-k model for a borehole in fold k), the **final** classifier for the holdout boreholes (it never saw them). The logged soil symbol is not an input.
4. Marks the **11 external-holdout boreholes** of V3 (checked against V3's lock) and the development boreholes (those in V3's blockCV folds). Boreholes in neither have no V3 fold and are left out (their number is reported).
5. **Layers:** averages the samples of each borehole in 1.5 m layers (`settings.toml → [layers]`): the target and P(Clay) are layer means.
6. Saves one file per target in `data/processed/` (not in git) and summary tables without borehole codes in `outputs/01_data/`.

No gap filling: these are point values. SPT refusals and "non-plastic" entries were blanked in the CSV export, so they are missing, not zero (see the source review, §5).

**Run time:** under 1 minute.

```python
import json, sys, time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

sys.path.insert(0, str(next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "settings.toml").exists())))
import helpers as H

S = H.load_settings()
T0 = time.time()
print("QUICK TEST RUN - not thesis results" if S["run"]["quick"] else "FULL RUN", "| V3 folder:", S["v3_root"])
```

## 1. Samples per target

```python
records = H.load_downtown(S)
holdout = H.load_v3_holdout(S)
folds = H.load_folds(S, H.block_sizes(S)[0])
fold_models, final_clf, clf_info = H.load_classifiers(S)
print("Classifier:", clf_info, "| out-of-fold models:", sorted(fold_models))
P, O = H.processed_dir(S), H.out_dir(S, "01_data")
rows, info = [], {}
for t in S["cv"]["targets"]:
    smp, inf = H.make_samples(records, t, S)
    smp["set"] = H.assign_sets(smp, holdout.index, folds)
    smp["p_clay"] = H.add_p_clay(smp, folds, fold_models, final_clf, S)
    smp.to_csv(P / f"raw_samples_{t}.csv", index=False)
    lay = H.to_layers(smp, S)
    lay["set"] = H.assign_sets(lay, holdout.index, folds)
    lay.to_csv(P / f"samples_{t}.csv", index=False)          # the LAYER table used by notebooks 02-04
    info[t] = {**inf, "layers": len(lay)}
    for s, g in lay.groupby("set"):
        rows.append({"target": t, "set": s, "boreholes": g.BH_Code.nunique(), "layers": len(g), "samples": int(g.n_samples.sum()),
                     "mean_p_clay": g.p_clay.mean(), "logged_clay_share": g.logged_clay_share.mean(),
                     "mean": g.y.mean(), "sd": g.y.std(), "min": g.y.min(), "median": g.y.median(), "max": g.y.max()})
counts = pd.DataFrame(rows)
counts.to_csv(O / "layer_counts.csv", index=False)
display(pd.DataFrame(info).T)
display(counts.round(2))

# how well the predicted P(Clay) matches the logged class (development samples with a Clay/Sand symbol)
from sklearn.metrics import roc_auc_score
chk = pd.concat([pd.read_csv(P / f"raw_samples_{t}.csv") for t in S["cv"]["targets"]]).drop_duplicates(["BH_Code", "depth_m"])
chk = chk[chk.soil_clay.notna()]
for s_, g in chk.groupby("set"):
    if g.soil_clay.nunique() == 2:
        print(f"{s_}: AUC of P(Clay) vs logged class = {roc_auc_score(g.soil_clay, g.p_clay):.3f} ({len(g)} samples)")
```

## 2. Profiles with depth (development set)
Mean and interquartile range per 1 m slice. A model that only knows depth already captures this trend; the spatial models must beat it (notebook 03).

```python
ts = S["cv"]["targets"]
fig, axes = plt.subplots(1, len(ts), figsize=(3.2 * len(ts), 4.5), sharey=True)
prof = []
for ax, t in zip(np.atleast_1d(axes), ts):
    d = pd.read_csv(P / f"raw_samples_{t}.csv")
    d = d[d.set == "development"]
    g = d.groupby(np.floor(d.depth_m)).y
    p = pd.DataFrame({"n": g.size(), "mean": g.mean(), "q25": g.quantile(0.25), "q75": g.quantile(0.75)})
    prof.append(p.assign(target=t))
    ax.fill_betweenx(p.index + 0.5, p.q25, p.q75, alpha=0.3)
    ax.plot(p["mean"], p.index + 0.5, marker="o")
    ax.set_title(t); ax.set_xlabel(S["targets"][t]["unit"])
np.atleast_1d(axes)[0].invert_yaxis(); np.atleast_1d(axes)[0].set_ylabel("depth (m)")
plt.tight_layout(); plt.savefig(O / "profiles_by_depth.png", dpi=200); plt.show()
pd.concat(prof).rename_axis("depth_from_m").to_csv(O / "profiles_by_depth.csv")
```

## 3. Value by logged soil class (development set)
How strongly each parameter follows the logged Clay/Sand class (for information; the model input is the predicted P(Clay)).

```python
grp = {**{c: "Clay" for c in ["CL", "CH"]}, **{c: "Sand" for c in ["SC", "SM", "SP"]}}
rows = []
for t in ts:
    d = pd.read_csv(P / f"raw_samples_{t}.csv")
    d = d[d.set == "development"].assign(group=lambda x: x.USCS.map(grp).fillna("other / missing"))
    rows.append(d.groupby("group").y.describe().assign(target=t))
by_uscs = pd.concat(rows).reset_index()
by_uscs.to_csv(O / "values_by_uscs_group.csv", index=False)
display(by_uscs.round(2))
```
